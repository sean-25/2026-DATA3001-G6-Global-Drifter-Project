```python
import pandas as pd
import numpy as np
```

### RONI EDA

```python
# --- Load RONI ---
RONI_PATH = "../data/RONI.ascii.txt"

inputFile = pd.read_csv(RONI_PATH, sep=r"\s+")
```

```python
# Each 3-month season -> date of its centre month (DJF 1950 -> 1950-01-01)
# reformatting for drifter data comparison

seas_to_month = {"DJF":1,"JFM":2,"FMA":3,"MAM":4,"AMJ":5,"MJJ":6,
                 "JJA":7,"JAS":8,"ASO":9,"SON":10,"OND":11,"NDJ":12}
inputFile["time"] = pd.to_datetime(dict(year=inputFile["YR"],
                                  month=inputFile["SEAS"].map(seas_to_month), day=1))

roni = inputFile.set_index("time")["ANOM"].rename("roni")
```

```python
roni.to_frame()
```

```python
roni.plot()
```

```python
# ENSO phase (NOAA thresholds: <= -0.5 La Niña, >= +0.5 El Niño)
roni_phase = pd.Series(
    np.select([roni <= -0.5, roni >= 0.5], ["La Niña", "El Niño"], default="Neutral"),
    index=roni.index, name="enso")
```

```python
# --- Sanity checks ---
assert roni.index.is_unique and roni.notna().all()
print(roni.index.min().date(), "to", roni.index.max().date(), f"({len(roni)} months)")
print(roni_phase.value_counts())
```

```python
start, end = "1990-01", "2022-10"
print(roni_phase[start:end].value_counts())
print(len(roni_phase[start:end]), "months of overlap")
```

```python
# --- Load region observation data ---
df = pd.read_parquet("../data/kuroshio_obs.parquet")
```

```python
df.info()
df.describe()
```

```python
bad_sst = (df["sst"] < 270) | (df["sst"] > 310)
print(bad_sst.sum(), "implausible SST values")
df.loc[bad_sst, "sst"] = np.nan            # blank them out, keep the velocity data
df["sst_c"] = df["sst"] - 273.15           # Celsius column for plotting
```

```python
df["drogue_status"].value_counts(dropna=False)
```

Data check (Kuroshio box, 115–145°E, 15–40°N). The extracted subset contains 4,630,468 hourly observations, with all positions inside the box and velocities in a physically realistic range (about ±2.8 m/s). In-box sampling begins in March 1990 rather than at the start of the GDP record (October 1987), and the median observation date is mid-2012. Half the data therefore comes from the last decade, which will weight any time-based composites (including ENSO phases) towards recent events. SST is stored in Kelvin; about 218k observations have no SST, and a further 10,042 (about 0.2%) had physically implausible values and were dismissed using a 270–310 K plausibility range, which is about -3 to 36 degrees celsius.

```python
# --- SST sanity checks ---
print(df["sst"].notna().sum())
print(df["sst"].min(), df["sst"].max()) 
```

```python
# Full subset (drogued + undrogued): use for SST
# Drogued only: use for current velocity (ve, vn)
df_drogued = df[df["drogue_status"]].copy()

print("SST analysis:     ", df["sst"].notna().sum(), "observations with valid SST")
print("Velocity analysis:", len(df_drogued), "drogued observations")
```

```python
# --- Tag every observation with its month's RONI value and phase ---
# Round each timestamp down to the 1st of its month to match the RONI index
month = pd.to_datetime(df["time"]).dt.to_period("M").dt.to_timestamp()
df["roni"] = month.map(roni)          # numeric RONI value
df["enso"] = month.map(roni_phase)    # "El Niño" / "La Niña" / "Neutral"
assert df["roni"].notna().all()       # every month 1990–2022 is in RONI, so none missing

# --- Recreate the drogued subset so it carries roni/enso too ---
df_drogued = df[df["drogue_status"]].copy()

# --- How much data each ENSO phase has ---
def phase_summary(d, value_col):
    s = d.groupby("enso").agg(n_obs=(value_col, "count"),     # non-missing values
                              n_drifters=("id", "nunique"))   # distinct drifters
    s["share_obs"] = (s["n_obs"] / s["n_obs"].sum()).round(3)
    return s

print("Velocity (drogued):\n", phase_summary(df_drogued, "ve"), "\n")
print("SST (all drifters):\n", phase_summary(df, "sst"))
```

```python
# Cleaned + RONI-tagged Kuroshio subset (all drifters; drogue filter applied in analysis)
df.to_parquet("../data/kuroshio_obs_clean.parquet")
```
